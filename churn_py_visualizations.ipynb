```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(
    data=df,
    x='Age',
    hue='Exited',
    bins=30,
    palette={0: '#1D9E75', 1: '#D85A30'}
)
plt.title('Age distribution — stayed vs churned')
plt.xlabel('Customer age')
plt.ylabel('Number of customers')
plt.legend(['Stayed', 'Churned'])
plt.savefig('chart3_age_churn.png', dpi=150)
plt.show()
```

```python
number_cols = ['CreditScore', 'Age', 'Tenure',
               'Balance', 'NumOfProducts',
               'IsActiveMember', 'EstimatedSalary', 'Exited']

plt.figure(figsize=(9, 6))
sns.heatmap(
    df[number_cols].corr(),
    annot=True,      
    fmt='.2f',       
    cmap='RdYlGn',   
    center=0
)
plt.title('Correlation heatmap — what drives churn?')
plt.savefig('chart4_heatmap.png', dpi=150)
plt.show()
```
